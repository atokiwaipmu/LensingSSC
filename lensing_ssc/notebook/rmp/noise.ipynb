```python
# Process RIP data
for box_type in box_types:
    for zs in zs_list:
        if (box_type, zs) not in combined_data_RIP:
            print(f"No RIP data for {box_type} box at z={zs}")
            continue

        for (ngal, sl), stats_dict in combined_data_RIP[(box_type, zs)].items():
            # Add to the list for correlation analysis
            if (box_type, zs) not in all_stats_RIP: all_stats_RIP[(box_type, zs)] = {}
            if (ngal, sl) not in all_stats_RIP[(box_type, zs)]: all_stats_RIP[(box_type, zs)][(ngal, sl)] = {}
            
            for stat_type, stat_arrays in stats_dict.items():
                if len(stat_arrays) == 0:
                    continue
                    
                # Stack arrays if multiple simulations available
                if isinstance(stat_arrays[0], np.ndarray):
                    stacked = np.vstack(stat_arrays)  # Shape: (n_sims, n_patches, n_bins)
                    
                    # Calculate mean and std across simulations for each patch and bin
                    mean_stats = np.mean(stacked, axis=0)  # Shape: (n_patches, n_bins)
                    std_stats = np.std(stacked, axis=0)    # Shape: (n_patches, n_bins)
                    
                    # Initialize dictionaries if needed
                    if (box_type, zs) not in means_RIP: means_RIP[(box_type, zs)] = {}
                    if (box_type, zs) not in stds_RIP: stds_RIP[(box_type, zs)] = {}
                    if (ngal, sl) not in means_RIP[(box_type, zs)]: means_RIP[(box_type, zs)][(ngal, sl)] = {}
                    if (ngal, sl) not in stds_RIP[(box_type, zs)]: stds_RIP[(box_type, zs)][(ngal, sl)] = {}
                    
                    # Store mean and std statistics
                    means_RIP[(box_type, zs)][(ngal, sl)][stat_type] = mean_stats
                    stds_RIP[(box_type, zs)][(ngal, sl)][stat_type] = std_stats
                    all_stats_RIP[(box_type, zs)][(ngal, sl)][stat_type] = stacked
                    
                    # Calculate correlation and covariance matrices
                    covariance_matrix = np.cov(stacked, rowvar=False)
                    correlation_matrix = np.corrcoef(stacked, rowvar=False)
                    
                    # Initialize dictionaries if needed
                    if (box_type, zs) not in covariances_RIP: covariances_RIP[(box_type, zs)] = {}
                    if (box_type, zs) not in correlations_RIP: correlations_RIP[(box_type, zs)] = {}
                    if (ngal, sl) not in covariances_RIP[(box_type, zs)]: covariances_RIP[(box_type, zs)][(ngal, sl)] = {}
                    if (ngal, sl) not in correlations_RIP[(box_type, zs)]: correlations_RIP[(box_type, zs)][(ngal, sl)] = {}
                    
                    # Store covariance and correlation matrices
                    covariances_RIP[(box_type, zs)][(ngal, sl)][stat_type] = covariance_matrix
                    correlations_RIP[(box_type, zs)][(ngal, sl)][stat_type] = correlation_matrix

print(f"Processed RIP statistics for {len(means_RIP)} box/redshift combinations")
```

### RMP

```python
import numpy as np

def cov_to_corr(C):
    d = np.sqrt(np.diag(C))
    with np.errstate(invalid='ignore', divide='ignore'):
        R = C / np.outer(d, d)
    R[~np.isfinite(R)] = 0.0
    return R

def jackknife_over_units(X):
    """
    X: (U, n_bins) array where each row is one jackknife unit (one patch from one realization).
    Returns mean, covariance, correlation, and per-bin errors of the mean estimator.
    """
    U, p = X.shape
    mu = X.mean(axis=0)                                    # full estimator
    # leave-one-out means using the closed form for the mean
    T = (U * mu - X) / (U - 1)                             # (U, p)
    Tbar = T.mean(axis=0)                                  # equals mu for the mean estimator
    diffs = T - Tbar
    C_jk = (U - 1) / U * (diffs.T @ diffs)                 # (p, p)
    R_jk = cov_to_corr(C_jk)
    err_jk = np.sqrt(np.diag(C_jk))
    return mu, C_jk, R_jk, err_jk

def jackknife_by_realization(stacked):
    """
    Block-JK: leave out all patches from one realization at a time.
    stacked: (n_sims, n_patches, n_bins)
    Returns mean, covariance, correlation, and errors for the mean estimator.
    """
    S, P, p = stacked.shape
    total_sum = stacked.sum(axis=(0,1))                    # (p,)
    mu = total_sum / (S * P)
    sum_per_sim = stacked.sum(axis=1)                      # (S, p), sums over patches within each realization
    # leave-one-realization-out means
    T = (total_sum[None, :] - sum_per_sim) / ((S - 1) * P) # (S, p)
    Tbar = T.mean(axis=0)                                  # equals mu
    diffs = T - Tbar
    C_jk = (S - 1) / S * (diffs.T @ diffs)
    R_jk = cov_to_corr(C_jk)
    err_jk = np.sqrt(np.diag(C_jk))
    return mu, C_jk, R_jk, err_jk
```

```python
means_RMP = {}
stds_RMP = {}
correlations_RMP = {}
covariances_RMP = {}
jk_errors_RMP = {}
all_stats_RMP = {}
```

```python
# Process RMP data
for box_type in box_types:
    for zs in zs_list:
        if (box_type, zs) not in combined_data_RMP:
            print(f"No RMP data for {box_type} box at z={zs}")
            continue

        for (ngal, sl), stats_dict in combined_data_RMP[(box_type, zs)].items():
            # Add to the list for correlation analysis
            if (box_type, zs) not in all_stats_RMP: all_stats_RMP[(box_type, zs)] = {}
            if (ngal, sl) not in all_stats_RMP[(box_type, zs)]: all_stats_RMP[(box_type, zs)][(ngal, sl)] = {}
            
            for stat_type, stat_arrays in stats_dict.items():
                if len(stat_arrays) == 0:
                    continue
                    
                # Stack arrays if multiple simulations available
                if isinstance(stat_arrays[0], np.ndarray):
                    stacked = np.vstack(stat_arrays)  # Shape: (n_sims * n_patches, n_bins)
                    mu, C_jk, R_jk, err_jk = jackknife_over_units(stacked)
                    
                    # Initialize dictionaries if needed
                    if (box_type, zs) not in means_RMP: means_RMP[(box_type, zs)] = {}
                    if (box_type, zs) not in covariances_RMP: covariances_RMP[(box_type, zs)] = {}
                    if (box_type, zs) not in correlations_RMP: correlations_RMP[(box_type, zs)] = {}
                    if (box_type, zs) not in jk_errors_RMP: jk_errors_RMP[(box_type, zs)] = {}

                    if (ngal, sl) not in means_RMP[(box_type, zs)]: means_RMP[(box_type, zs)][(ngal, sl)] = {}
                    if (ngal, sl) not in covariances_RMP[(box_type, zs)]: covariances_RMP[(box_type, zs)][(ngal, sl)] = {}
                    if (ngal, sl) not in correlations_RMP[(box_type, zs)]: correlations_RMP[(box_type, zs)][(ngal, sl)] = {}
                    if (ngal, sl) not in jk_errors_RMP[(box_type, zs)]: jk_errors_RMP[(box_type, zs)][(ngal, sl)] = {}

                    means_RMP[(box_type, zs)][(ngal, sl)][stat_type] = mu
                    covariances_RMP[(box_type, zs)][(ngal, sl)][stat_type] = C_jk
                    correlations_RMP[(box_type, zs)][(ngal, sl)][stat_type] = R_jk
                    jk_errors_RMP[(box_type, zs)][(ngal, sl)][stat_type] = err_jk

print(f"Processed RMP statistics for {len(means_RMP)} box/redshift combinations")
```

## average cov/corr

```python
def cal_corr_avg(corr):
    # Exclude diagonal and NaN values, then calculate the average
    mask = ~np.eye(corr.shape[0], dtype=bool) & ~np.isnan(corr)
    return np.nanmean(corr[mask])

def cal_corr_offdiag_avg(M):
    # average off-diagonal entries (NaNs ignored)
    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    return np.nanmean(vals[np.isfinite(vals)])

def safe_mean(arr):
    arr_no_inf = np.where(np.isinf(arr), np.nan, arr)
    return np.nanmean(arr_no_inf)

def nan_divide(num, den):
    out = np.full_like(num, np.nan, dtype=float)
    with np.errstate(invalid='ignore', divide='ignore'):
        np.divide(num, den, out=out, where=(den != 0))
    return out
```

```python
# Initialize dictionaries to store average ratios
avgs_cov = {}
avgs_corr = {}

# Number of bins for each statistic (used for excluding edge bins)
nbin = len(nu)
except_indices=[(1, nbin-1), (1, nbin-1), (1, nbin-1), (1, nbin-1), (1, nbin-1), (1, nbin-1)]

# Process for each noise level
for ngal_i in ngal_list:
    ngal = f"ngal_{ngal_i}"
    avgs_cov[ngal] = {}
    avgs_corr[ngal] = {}
    
    # Process for each redshift
    for i, zs_i in enumerate(zs_list):
        # Process ell-binned statistics (power spectrum, bispectra)
        for label in labels_ell[:1]:  # Just power spectra
            # Initialize lists if not already done
            if label not in avgs_cov[ngal]:
                avgs_cov[ngal][label] = []
            if label not in avgs_corr[ngal]:
                avgs_corr[ngal][label] = []
            
            # Get covariance and correlation matrices
            if (('bigbox', zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('bigbox', zs_i)] and
                ('tiled', zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('tiled', zs_i)]):
                if label in covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)] and label in covariances_RMP[('tiled', zs_i)][(ngal, sl_default)]:
                    # Calculate ratio of covariance matrices
                    cov_ratio = covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)][label] / covariances_RMP[('tiled', zs_i)][(ngal, sl_default)][label]
                    # Calculate ratio of correlation matrices
                    corr_ratio = correlations_RMP[('bigbox', zs_i)][(ngal, sl_default)][label] / correlations_RMP[('tiled', zs_i)][(ngal, sl_default)][label]
                    # Calculate mean of ratio (excluding NaN and inf values)
                    avgs_cov[ngal][label].append(safe_mean(cov_ratio))
                    avgs_corr[ngal][label].append(cal_corr_avg(corr_ratio))
        
        # Process nu-binned statistics (PDF, peaks, minima, MFs)
        for j, label in enumerate(labels_nu):
            # Initialize lists if not already done
            if label not in avgs_cov[ngal]:
                avgs_cov[ngal][label] = []
            if label not in avgs_corr[ngal]:
                avgs_corr[ngal][label] = []
            
            # Define range of bins to use (exclude edges)
            rmin, rmax = except_indices[j]
            if (('bigbox', zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('bigbox', zs_i)] and
                ('tiled', zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('tiled', zs_i)]):
                if label in covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)] and label in covariances_RMP[('tiled', zs_i)][(ngal, sl_default)]:
                    # Get covariance matrices
                    bigbox_cov = covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]
                    tiled_cov = covariances_RMP[('tiled', zs_i)][(ngal, sl_default)][label]
                    
                    # Get correlation matrices
                    bigbox_corr = correlations_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]
                    tiled_corr = correlations_RMP[('tiled', zs_i)][(ngal, sl_default)][label]
                    # Calculate ratios
                    cov_ratio = bigbox_cov / tiled_cov
                    corr_ratio = bigbox_corr / tiled_corr
                    
                    # Calculate means of ratios in the specified range
                    avgs_cov[ngal][label].append(safe_mean(cov_ratio[rmin:rmax, rmin:rmax]))
                    avgs_corr[ngal][label].append(cal_corr_avg(corr_ratio[rmin:rmax, rmin:rmax]))
```

```python
avgs_ssc_covfrac = {}   # <ΔC / C_tiled> mean on sub-block
avgs_ssc_corrlike = {}  # <offdiag of ΔC normalized to diag(C_tiled)>

# Number of bins for nu-binned stats (for edge exclusion)
nbin = len(nu)
except_indices = [(1, nbin-1), (1, nbin-1), (1, nbin-1), (1, nbin-1), (1, nbin-1), (1, nbin-1)]

for ngal_i in ngal_list:
    ngal = f"ngal_{ngal_i}"
    avgs_cov.setdefault(ngal, {})
    avgs_corr.setdefault(ngal, {})
    avgs_ssc_covfrac.setdefault(ngal, {})
    avgs_ssc_corrlike.setdefault(ngal, {})

    for zs_i in zs_list:
        # ---------- ℓ-binned (example: power spectrum only here)
        for label in labels_ell[:1]:
            avgs_cov[ngal].setdefault(label, [])
            avgs_corr[ngal].setdefault(label, [])
            avgs_ssc_covfrac[ngal].setdefault(label, [])
            avgs_ssc_corrlike[ngal].setdefault(label, [])

            ok = (
                ('bigbox', zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('bigbox', zs_i)] and
                ('tiled',  zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('tiled',  zs_i)] and
                label in covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)] and
                label in covariances_RMP[('tiled',  zs_i)][(ngal, sl_default)]
            )
            if not ok: 
                continue

            # JK covariances and correlations (of the mean estimator)
            C_big = covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]
            C_til = covariances_RMP[('tiled',  zs_i)][(ngal, sl_default)][label]
            R_big = correlations_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]
            R_til = correlations_RMP[('tiled',  zs_i)][(ngal, sl_default)][label]

            # Your original ratios (keep if you still want them)
            cov_ratio  = nan_divide(C_big, C_til)
            corr_ratio = nan_divide(R_big, R_til)
            avgs_cov[ngal][label].append(safe_mean(cov_ratio))
            avgs_corr[ngal][label].append(cal_corr_offdiag_avg(corr_ratio))

            # ---- SSC: excess covariance and its normalizations
            dC = C_til - C_big                                     # ΔC_SSC
            # (i) element-wise SSC fraction in covariance space
            ssc_covfrac = nan_divide(dC, C_til)                    # ΔC / C_tiled
            avgs_ssc_covfrac[ngal][label].append(safe_mean(ssc_covfrac))

            # (ii) "correlation-like" SSC pattern (not a true corr matrix)
            denom = np.sqrt(np.outer(np.diag(C_til), np.diag(C_til)))
            R_ssc_like = nan_divide(dC, denom)
            avgs_ssc_corrlike[ngal][label].append(cal_corr_offdiag_avg(R_ssc_like))

        # ---------- ν-binned (PDF/peaks/minima/MFs...)
        for j, label in enumerate(labels_nu):
            avgs_cov[ngal].setdefault(label, [])
            avgs_corr[ngal].setdefault(label, [])
            avgs_ssc_covfrac[ngal].setdefault(label, [])
            avgs_ssc_corrlike[ngal].setdefault(label, [])

            rmin, rmax = except_indices[j]

            ok = (
                ('bigbox', zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('bigbox', zs_i)] and
                ('tiled',  zs_i) in covariances_RMP and (ngal, sl_default) in covariances_RMP[('tiled',  zs_i)] and
                label in covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)] and
                label in covariances_RMP[('tiled',  zs_i)][(ngal, sl_default)]
            )
            if not ok:
                continue

            C_big = covariances_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]
            C_til = covariances_RMP[('tiled',  zs_i)][(ngal, sl_default)][label]
            R_big = correlations_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]
            R_til = correlations_RMP[('tiled',  zs_i)][(ngal, sl_default)][label]

            # restrict to inner bins (exclude edges)
            slc = np.s_[rmin:rmax, rmin:rmax]
            cov_ratio  = nan_divide(C_big[slc], C_til[slc])
            corr_ratio = nan_divide(R_big[slc], R_til[slc])

            avgs_cov[ngal][label].append(safe_mean(cov_ratio))
            avgs_corr[ngal][label].append(cal_corr_offdiag_avg(corr_ratio))

            # SSC targets
            dC = C_til[slc] - C_big[slc]
            ssc_covfrac = nan_divide(dC, C_til[slc])  # ΔC / C_tiled on the sub-block
            avgs_ssc_covfrac[ngal][label].append(safe_mean(ssc_covfrac))

            denom = np.sqrt(np.outer(np.diag(C_til[slc]), np.diag(C_til[slc])))
            R_ssc_like = nan_divide(dC, denom)
            avgs_ssc_corrlike[ngal][label].append(cal_corr_offdiag_avg(R_ssc_like))
```

```python
# ------- helpers -------
def nan_divide(a, b):
    out = np.full_like(a, np.nan, dtype=float)
    with np.errstate(divide='ignore', invalid='ignore'):
        np.divide(a, b, out=out, where=(b!=0))
    return out

def diag_var(C):
    # robust diagonal extraction; ensures 1D (n_bins,)
    d = np.diag(C).astype(float)
    d[~np.isfinite(d)] = np.nan
    return d

def auto_ratio_ylim(r, pad=0.1, clamp=(0.5, 2.0)):
    # symmetric around 1 based on max deviation in finite entries
    rfin = r[np.isfinite(r)]
    if rfin.size == 0:
        return (0.8, 1.2)
    dev = np.nanmax(np.abs(rfin - 1.0))
    lo, hi = 1 - (1+pad)*dev, 1 + (1+pad)*dev
    lo, hi = max(lo, clamp[0]), min(hi, clamp[1])
    # ensure a minimum span
    if hi - lo < 0.1: 
        lo, hi = 0.95, 1.05
    return (lo, hi)
```

```python
def cov_to_corr(C):
    d = np.sqrt(np.diag(C))
    with np.errstate(divide='ignore', invalid='ignore'):
        R = C / np.outer(d, d)
    R[~np.isfinite(R)] = 0.0
    return R

def jackknife_over_units(X):
    """
    X: (U, p) array; rows=units, cols=bins/stat-bins.
    Returns:
      mu       : (p,) mean over units
      C_jk     : (p, p) jackknife covariance of the mean estimator
      R_jk     : (p, p) correlation matrix from C_jk
      err_mean : (p,) sqrt(diag(C_jk)) -- 1σ on the mean estimator
    """
    X = np.asarray(X)
    if X.ndim != 2:
        X = X.reshape(X.shape[0], -1)  # safety if (S, P, p) slipped in
    U, p = X.shape
    mu = X.mean(axis=0)
    T = (U * mu - X) / (U - 1)                    # leave-1-out means, shape (U, p)
    Tbar = T.mean(axis=0)                          # equals mu numerically
    diffs = T - Tbar
    C_jk = (U - 1) / U * (diffs.T @ diffs)
    R_jk = cov_to_corr(C_jk)
    err = np.sqrt(np.diag(C_jk))
    return mu, C_jk, R_jk, err

def jk_var_with_error(X, compute_var_err=True):
    """
    Returns:
      var_diag : diag(C_jk)  [variance of the mean estimator per bin]
      var_err  : jackknife SE of var_diag (same shape), or None if disabled
    """
    _, C_jk, _, _ = jackknife_over_units(X)
    var_diag = np.diag(C_jk).copy()

    if not compute_var_err:
        return var_diag, None

    # "JK of JK": recompute diag(C_jk) leaving one *unit* out each time.
    X = np.asarray(X)
    U = X.shape[0]
    var_loolist = []
    for u in range(U):
        Xm = np.delete(X, u, axis=0)
        _, C_u, _, _ = jackknife_over_units(Xm)
        var_loolist.append(np.diag(C_u))
    V = np.vstack(var_loolist)               # (U, p)
    Vbar = V.mean(axis=0)
    var_err = np.sqrt((U - 1) / U * np.sum((V - Vbar)**2, axis=0))
    return var_diag, var_err

def ratio_error_delta(A, Aerr, B, Berr):
    """
    Delta-method error on R = A/B, assuming A and B independent.
    """
    with np.errstate(divide='ignore', invalid='ignore'):
        R = A / B
        rel2 = np.zeros_like(R, dtype=float)
        rel2 += np.where(A > 0, (Aerr / A)**2, 0.0)
        rel2 += np.where(B > 0, (Berr / B)**2, 0.0)
        Rerr = np.abs(R) * np.sqrt(rel2)
    # clean any non-finite (e.g., zeros)
    R[~np.isfinite(R)] = np.nan
    Rerr[~np.isfinite(Rerr)] = np.nan
    return R, Rerr
```

```python
colors = ["tab:blue", "tab:orange", "tab:green", "tab:red", "tab:purple"]
survey_info = {'0':'Noiseless', '7':'DES', '15':'HSC', '30':'LSST', '50':'Roman'}
labels_ell = ["power_spectra", "equilateral", "isosceles", "squeezed"]
titles_ell = [r"$C_{\ell}$", r"$B_{\ell}^{(\mathrm{eq})}$", r"$B_{\ell}^{(\mathrm{iso})}$", r"$B_{\ell}^{(\mathrm{sq})}$"]
fontsize = 16

lmin, lmax = 300, 3000
mranges_ell = [(0.95, 1.25), (0.9, 1.2), (0.9, 1.2), (0.9, 1.2)]  # ratio panel ranges

linestyles = {"bigbox":"-", "tiled":"--"}
markers_ds = {"bigbox":"o", "tiled":"s"}  # distinguish boxes by marker

# Toggle this to speed up if units are huge
COMPUTE_VAR_ERR = True

fig = plt.figure(figsize=(14, 3 * len(labels_ell)))
fig.suptitle(
    r"Jackknife variance of the mean, $\mathrm{diag}\,C_{\mathrm{JK}}$, "
    r"and ratio $R \equiv \mathrm{diag}\,C^{\mathrm{big}}_{\mathrm{JK}}/\mathrm{diag}\,C^{\mathrm{til}}_{\mathrm{JK}}$ "
    r"for Fourier-space statistics at $z_s=0.5$ (Left) and $z_s=1.5$ (Right)",
    fontsize=16, y=0.94
)

gs_master = GridSpec(nrows=len(labels_ell), ncols=2, hspace=0.25, wspace=0.25)
gs0 = [GridSpecFromSubplotSpec(2, 1, subplot_spec=gs_master[i, 0], height_ratios=[3, 1], hspace=0.08) for i in range(len(labels_ell))]
gs1 = [GridSpecFromSubplotSpec(2, 1, subplot_spec=gs_master[i, 1], height_ratios=[3, 1], hspace=0.08) for i in range(len(labels_ell))]

ax_upper0 = [fig.add_subplot(gs0[i][0]) for i in range(len(labels_ell))]
ax_upper1 = [fig.add_subplot(gs1[i][0]) for i in range(len(labels_ell))]
ax_lower0 = [fig.add_subplot(gs0[i][1]) for i in range(len(labels_ell))]
ax_lower1 = [fig.add_subplot(gs1[i][1]) for i in range(len(labels_ell))]

# ------------- MAIN LOOP (ell-binned)
for zs_i, axsU, axsL in [(0.5, ax_upper0, ax_lower0), (1.5, ax_upper1, ax_lower1)]:
    for j, label in enumerate(labels_ell):
        for k, ngal_i in enumerate(ngal_list):
            ngal = f"ngal_{ngal_i}"

            # fetch unit-level arrays (units, nbins) for each box
            ok_big = (('bigbox', zs_i) in all_stats_RMP and
                      (ngal, sl_default) in all_stats_RMP[('bigbox', zs_i)] and
                      label in all_stats_RMP[('bigbox', zs_i)][(ngal, sl_default)])
            ok_til = (('tiled',  zs_i) in all_stats_RMP and
                      (ngal, sl_default) in all_stats_RMP[('tiled',  zs_i)] and
                      label in all_stats_RMP[('tiled',  zs_i)][(ngal, sl_default)])

            if not (ok_big and ok_til):
                continue

            X_big = all_stats_RMP[('bigbox', zs_i)][(ngal, sl_default)][label]  # (U_b, B)
            X_til = all_stats_RMP[('tiled',  zs_i)][(ngal, sl_default)][label]  # (U_t, B)

            # JK variance of the mean (per bin) + error on that variance
            v_big, v_big_err = jk_var_with_error(X_big, compute_var_err=COMPUTE_VAR_ERR)
            v_til, v_til_err = jk_var_with_error(X_til, compute_var_err=COMPUTE_VAR_ERR)

            # ratio (with delta-method error; assumes independence between suites)
            R, Rerr = ratio_error_delta(v_big, v_big_err if v_big_err is not None else np.zeros_like(v_big),
                                        v_til, v_til_err if v_til_err is not None else np.zeros_like(v_til))

            # choose axis set by redshift
            axU = axsU[j]
            axL = axsL[j]

            # ---- UPPER: variances with error bars (log y)
            # plot bigbox and tiled for this noise level
            for ds_name, v, verr in [("bigbox", v_big, v_big_err), ("tiled", v_til, v_til_err)]:
                axU.errorbar(
                    ell, v, yerr=verr, lw=1.5,
                    color=colors[k], linestyle=linestyles[ds_name], marker=markers_ds[ds_name],
                    markersize=3, capsize=2, capthick=1, alpha=0.9, label=None
                )

            # ---- LOWER: ratio with error bars (linear y around 1)
            axL.errorbar(
                ell, R, yerr=Rerr, lw=1.5, color=colors[k],
                linestyle='-', marker='.', markersize=3, capsize=2, capthick=1, alpha=0.9
            )

# ---------- Cosmetics shared
for ax in ax_upper0 + ax_upper1 + ax_lower0 + ax_lower1:
    ax.set_xscale('log')
    ax.set_xlim(lmin, lmax)
    ax.minorticks_on()
    ax.tick_params(axis='both', which='major', length=6, labelsize=11)
    ax.tick_params(axis='both', which='minor', length=3)
    ax.set_xticks([300, 400, 600, 1000, 2000, 3000])
    ax.set_xticklabels(["300", "400", "600", "1000", "2000", "3000"])

for ax in ax_upper0 + ax_upper1:
    ax.set_yscale("log")
    ax.tick_params(labelbottom=False)

for i, ax in enumerate(ax_lower0 + ax_lower1):
    ax.axhline(1.0, color="black", linestyle="--", lw=1)
    if i == len(labels_ell) - 1 or i == 2*len(labels_ell) - 1:
        ax.set_xlabel(r"$\ell$", fontsize=fontsize)

# ratio panel ranges per stat
for i, ax in enumerate(ax_lower0):
    ax.set_ylim(mranges_ell[i][0], mranges_ell[i][1])
for i, ax in enumerate(ax_lower1):
    ax.set_ylim(mranges_ell[i][0], mranges_ell[i][1])

for i, ax in enumerate(ax_upper0):
    ax.set_ylabel(rf"$\mathrm{{diag}}\,C_{{\mathrm{{JK}}}}$: {titles_ell[i]}", fontsize=fontsize)
for i, ax in enumerate(ax_lower0):
    ax.set_ylabel(r"$R \equiv \mathrm{diag}\,C^{\mathrm{big}}_{\mathrm{JK}}/\mathrm{diag}\,C^{\mathrm{til}}_{\mathrm{JK}}$", fontsize=fontsize)
for i, ax in enumerate(ax_upper1):
    ax.set_ylabel(rf"$\mathrm{{diag}}\,C_{{\mathrm{{JK}}}}$: {titles_ell[i]}", fontsize=fontsize)
for i, ax in enumerate(ax_lower1):
    ax.set_ylabel(r"$R$", fontsize=fontsize)

# Legend: colors = noise levels; linestyle/marker = box type
custom_noise_lines = [Line2D([0],[0], color=colors[i], lw=2) for i in range(len(ngal_list))]
custom_box_lines   = [Line2D([0],[0], color="black", lw=2, linestyle=linestyles['bigbox'], marker=markers_ds['bigbox'], markersize=6),
                      Line2D([0],[0], color="black", lw=2, linestyle=linestyles['tiled'],  marker=markers_ds['tiled'],  markersize=6)]
legend_labels = [f"{survey_info[str(ngal_i)]}" for ngal_i in ngal_list] + ["Bigbox", "Tiled"]

fig.legend(custom_noise_lines + custom_box_lines, legend_labels,
           bbox_to_anchor=(0.5, 0.06), loc='upper center',
           ncol=len(ngal_list) + 2, fontsize=12)

plt.show()
fig.savefig(f"{savedir}/ell_main_noise_jk.pdf", bbox_inches='tight')
```
